# Explicación Drop Schema

Este script realiza las siguientes operaciones paso a paso:

1. **Obtener todos los catálogos disponibles**  
   catalogs_df = spark.sql("SHOW CATALOGS")

2. Filtrar catálogos cuyo nombre comienza con "capa"
  capa_catalogs = catalogs_df.filter(catalogs_df['catalog'].startswith("capa"))

3. Iterar sobre cada catálogo filtrado
  for row in capa_catalogs.collect():
    catalog = row['catalog']

4. Buscar el esquema llamado "tables" dentro de cada catálogo
  schemas_df = spark.sql(f"SHOW SCHEMAS IN {catalog}")
  if schemas_df.filter(schemas_df['databaseName'] == "tables").count() > 0:

5. Eliminar el esquema "tables" si existe
  spark.sql(f"DROP SCHEMA {catalog}.tables CASCADE")


```python
catalogs_df = spark.sql("SHOW CATALOGS")
capa_catalogs = catalogs_df.filter(catalogs_df['catalog'].startswith("capa"))

for row in capa_catalogs.collect():
    catalog = row['catalog']
    schemas_df = spark.sql(f"SHOW SCHEMAS IN {catalog}")
    if schemas_df.filter(schemas_df['databaseName'] == "tables").count() > 0:
        spark.sql(f"DROP SCHEMA {catalog}.tables CASCADE")
```
